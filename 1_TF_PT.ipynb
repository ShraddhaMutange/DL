```python
print("x - x : \n", x - x)
```

```python
print(x@tf.transpose(x)) # matrix multiplication and transpose
```

```python
result = tf.nn.softmax(x, axis=-1)
print(result)
```

## **Keras**

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
df = pd.read_csv("diabetes.csv")
```

```python
df.shape
```

```python
df.head(5)
```

```python
x = df.iloc[:,0:8]
print(type(x))
```

```python
print(x.shape)
```

```python
y = df.iloc[:,8]
print(y)
```

```python
model = Sequential()
```

```python
model.add(Dense(
        12,
        input_shape = (8,),
        activation='relu'
))
```

```python
model.add(Dense(5, activation='relu'))
```

```python
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(x,y, epochs=50, batch_size=5)
```

```python
model.evaluate(x,y)
```
