```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, Lambda
text = """
the cat sits on the mat
the cat likes the mat
the dog sits on the mat
the dog likes the cat
"""
words = text.lower().split()
print("Words : ",words)
vocab = sorted(set(words))
print("Vocabulary :\n",vocab)
word_to_id = {
    word : i
    for i,word in enumerate(vocab)
}

print("Word to id :")
print(word_to_id)
id_to_word = {
    i : word 
    for word,i in word_to_id.items()
}
print("id to Word :")
print(id_to_word)
encoded_words = [
    word_to_id[word]
    for word in words
]
print("Encoded Words :")
print(encoded_words)
window_size = 2
contexts = []
targets = []
for i in range(window_size, len(encoded_words)-window_size):
    left_context = encoded_words[ i - window_size : i]

    right_context = encoded_words[ i+1 : i+1+window_size]

    context = left_context + right_context

    target = encoded_words[i]

    contexts.append(context)
    targets.append(target)

contexts = np.array(contexts)
targets = np.array(targets)

print("Shape :")
print("Context : ",contexts.shape)
print("Target : ",targets.shape)
print("Training Samples :\n")

for i in range(5):
    context_words = [
        id_to_word[x]
        for x in contexts[i]
    ]

    target_words = id_to_word[targets[i]]

    print("Context :\n",context_words)
    print("Target :\n",target_words)
embedding_dim = 10
vocab_size = len(vocab)
model = Sequential([
    Embedding(
        input_dim = vocab_size,
        output_dim = embedding_dim),

    Lambda( lambda x : tf.reduce_mean(x,axis=1)),

    Dense(
        vocab_size,
        activation = "softmax")
])

model.summary()
model.compile(
    optimizer = "adam",
    loss = "sparse_categorical_crossentropy",
    metrics = ["accuracy"]
)
history = model.fit(
    contexts, targets,
    epochs = 100, verbose =1 )
predictions = model.predict(contexts)
predicted_ids = np.argmax(predictions,axis =1)
print("Predictions :\n")
for i in range(5):
    context_words = [
        id_to_word[x]
        for x in contexts[i]
    ]

    actual_word = id_to_word[targets[i]]

    predicted_word = id_to_word[predicted_ids[i]]

    print("Context : ",context_words)
    print("Actual : ",actual_word)
    print("Prediction : ",predicted_word)

embeddings = model.layers[0].get_weights()[0]

print("\nLearned word embeddings :")
print(embeddings)
```
